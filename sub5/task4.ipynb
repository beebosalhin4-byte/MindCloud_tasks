```python
import numpy as np

def polynomial_features(x, order):
    x = np.array(x)
    
    return np.column_stack([x**i for i in range(order + 1)])

x_input = input("Enter the x list (separated by spaces or commas): ")
x_data = [float(num) for num in x_input.replace(",", " ").split()]
order = int(input("Enter the Order: "))
matrix = polynomial_features(x_data, order)
print(matrix)
```
