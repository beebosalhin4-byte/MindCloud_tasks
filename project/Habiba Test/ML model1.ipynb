```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
import Features2 as f
import importlib
```

```python
importlib.reload(f)
audio=pd.read_csv("finalfeatures23.csv")
audio=audio.drop(audio.columns[0],axis="columns")
audio=audio.drop(columns=["File name","Path"])
target=audio["Category"]
input=audio.loc[:,"f0_mean":]
```

```python
model=RandomForestClassifier(n_estimators=300,max_depth=6, random_state= 41,
                             min_samples_leaf=10, min_samples_split=15, class_weight="balanced")
model.fit(input,target)
ypred0=model.predict(input)
```

```python
test=pd.read_csv("finalTest2.csv")
test=test.drop(columns=["Unnamed: 0"])
testin=test.loc[:,"f0_mean":]
testout=test["Category"]
model.score(testin,testout)
ypred1=model.predict(testin)
```

```python
from sklearn.metrics import confusion_matrix,classification_report
t=confusion_matrix(testout,ypred1)
print(classification_report(testout,ypred1))
print(t)
print(model.score(testin,testout))
```
