```python
def verify_feature_extraction(
    merged_df, atol=1e-2, rtol=1e-2, pass_threshold_pct=90.0
):
    # Mapping: (Extracted CSV Column Name -> Reference CSV Column Name)
    feature_pairs = {
        "mfcc mean": "MFCCs13Mean",
        "zcr_mean": "ZCR_Mean",
        "rmsmean": "RMS_Mean",
    }

    results = {}

    for col_ext, col_ref in feature_pairs.items():
        # Compute absolute difference
        diff = np.abs(merged_df[col_ext] - merged_df[col_ref])

        # Check floating-point proximity
        is_close = np.isclose(
            merged_df[col_ext], merged_df[col_ref], atol=atol, rtol=rtol
        )

        match_pct = (
            (is_close.sum() / len(merged_df)) * 100
            if len(merged_df) > 0
            else 0
        )

        results[col_ext] = {
            "ref_column": col_ref,
            "min_difference": diff.min(),
            "mean_difference": diff.mean(),
            "max_difference": diff.max(),
            "matching_samples": is_close.sum(),
            "total_samples": len(merged_df),
            "match_rate_pct": match_pct,
            "passed_validation": match_pct >= pass_threshold_pct,
        }

    # Determine overall status
    all_passed = all(
        metrics["passed_validation"] for metrics in results.values()
    )

    # Print Detailed Summary Report
    print(f"Total matching files compared: {len(merged_df)}\n")
    for col_ext, metrics in results.items():
        status = "PASSED" if metrics["passed_validation"] else "FAILED"
        print(
            f"--- Comparing '{col_ext}' vs '{metrics['ref_column']}' [{status}] ---"
        )
        print(f"Min Absolute Diff : {metrics['min_difference']:.6f}")
        print(f"Mean Absolute Diff: {metrics['mean_difference']:.6f}")
        print(f"Max Absolute Diff : {metrics['max_difference']:.6f}")
        print(
            f"Matches (within tol): {metrics['matching_samples']}/{metrics['total_samples']} ({metrics['match_rate_pct']:.2f}%)\n"
        )

    print("================ OVERALL VERDICT ================")
    if all_passed:
        print(
            "SUCCESS: Features match closely enough to assume feature extraction was successful!"
        )
    else:
        print(
            "FAILED: Differences are too large to assume successful feature extraction."
        )

    return merged_df, results
```
